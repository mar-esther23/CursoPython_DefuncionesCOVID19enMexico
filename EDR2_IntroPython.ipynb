## Valores, variables y tipos

Antes de empezar necesitamos algunas definiciones:

* Valor: la medida o atributo real
* Variable: una medida o atributo que puede cambiar

Por ejemplo, `5` es un valor, pero `a` es una variable que puede tener cualquier valor, como 5.

````python
a = 5
a
````

En Python existen diferentes tipos de datos, como números enteros, flotantes, texto y booleanos.

Los números enteros se escriben sin decimales, mientras que los flotantes tienen decimales. El texto se escribe entre comillas simples o dobles. Los booleanos son verdadero o falso.

* _integer_ (`int`): número entero
* _float_ (`float`): número decimal
* _string_ (`str`): cadena de texto
* _boolean_ (`bool`): verdadero o falso
* `None` y `np.nan`: cuando no existe

Para saber el tipo de un valor usamos la función `type()`

Por defecto, el notebook nos muestra la salida de las instrucciones en la última línea de la celda sin necesidad de usar la instrucción `print()`

Asignemos a la variable `a` el valor entero `5`

````python
a = 5
a, type(a)
````

Ahora, asignemos a la misma variable `a` el valor flotante `5.0`. Podemos ver que aunque la variable es la misma, el tipo cambia.

````python
a = 5.0
a, type(a)
````

También podemos asignar a `a` el texto `'cinco'`, marcando entre comillas que es un valor de texto.

Cabe destacar que aunque para un humano el valor es el mismo, para la computadora son tres valores diferentes con distintas características. Tanto `5` como `5.0` son números y se pueden en usar operaciones matemáticas, aunque se representan diferente en la memoria.
Por otro lado, `cinco` es un texto y se le aplican operaciones diferentes.

````python
a = 'cinco'
a, type(a)
````

Otro tipo de datos son los Booleanos, los cuales pueden ser Verdaderos o Falsos.

````python
a = False
a, type(a)
````

A veces necesitamos representar la falta de información: `None` se utiliza para representar cuando el valor "no existe".

````python
a = None
a, type(a)
````

Python incluye una serie de herramientas por default, pero algunas más especializadas se encuentran en bibliotecas especiales. Por ejemplo, `nan` o "not a number" es parte de biblioteca de operaciones matemáticas `numpy`.

````python
from numpy import nan
a = nan
a, type(a)
````

### Listas

Además, existen tipos que son colecciones de otros valores:
* `list`: lista ordenada
* `dict`: diccionario con llaves

Primero definiremos una lista, en este caso usaremos una nueva variable llamada `l`

````python
l = [1, 2, 3, 4, 5, 6, 7, 8, 9, 0]
l, type(l)
````

Podemos acceder a los elementos de las listas por su posición. Las listas empiezan de cero.

````python
l[0]
````

````python
l[3]
````

El usar números negativos en el índice nos regresa el elemento contando del final hacia el principio

````python
l[-1]
````

También podemos acceder a múltiples consecutivos elementos de la lista con el formato `inicio : final`.
Nota que no se incluye el último número.

````python
l[2:6]
````

Si no se marca el final en automático se pone el resto de la lista

````python
l[4:]
````

Las listas pueden contener valores de distintos tipos, incluyendo otras listas

````python
l = [1 , 2.0, 'tres', 4, ['cinco','five', '5','cinq', 5], 6, 'siete']
l, type(l)
````

## Flujo de un Programa en Python

El flujo de un programa en Python se refiere a la secuencia en la que las instrucciones son ejecutadas. Generalmente, un programa de Python se ejecuta de arriba hacia abajo, línea por línea. Sin embargo, este flujo puede ser modificado por:

*   **Condicionales (`if`/`elif`/`else`)**: Permiten que el programa tome diferentes caminos basados en si una condición es verdadera o falsa.
*   **Bucles (`for`/`while`)**: Permiten repetir un bloque de código múltiples veces. Un bucle `for` itera sobre una secuencia (como una lista), mientras que un bucle `while` se repite mientras una condición sea verdadera.
*   **Funciones**: Cuando se llama a una función, el flujo de ejecución salta a esa función, ejecuta el código dentro de ella y luego regresa al punto donde fue llamada.
*   **Manejo de Excepciones (`try`/`except`)**: Permite manejar errores que pueden ocurrir durante la ejecución, alterando el flujo normal del programa si se detecta un error.

> Nota: En un notebook, el flujo es aún más interactivo, ya que las celdas pueden ejecutarse individualmente en cualquier orden.

Los condicionales son estructuras de control que permiten al programa ejecutar diferentes bloques de código dependiendo de si una o más condiciones son verdaderas o falsas.

* **`if`**: Es el bloque condicional básico. Si la condición que le sigue es `True`, el código dentro del bloque `if` se ejecuta. Si es `False`, el programa omite este bloque y pasa a la siguiente parte.
* **`elif` (else if)**: Se utiliza después de un `if` si este es `False` para probar una nueva condición y, si es necesario, ejecutar el código correspondiente. Puede haber múltiples `elif`.
* **`else`**: Es el bloque final y opcional de una estructura condicional, el código se ejecuta si ninguna de las condiciones anteriores (`if` o `elif`) fue `True`.

````python
edad = 65
if edad < 18:
    print(f"Edad {edad}: menor de edad")
elif edad >= 60:
    print(f"Edad {edad}: adulto mayor")
else:
    print(f"Edad {edad}: adulto")
````

Los bucles `for` en Python sirven para iterar sobre una secuencia u objeto iterable (por ejemplo lista o diccionario). Esto significa que se ejecuta un bloque de código una vez por cada elemento en la secuencia. Sirven para automatizar tareas repetitivas y procesar colecciones de datos.

En general, un bucle tiene la forma:

```python
for item in secuencia:
    # Código a ejecutar en cada iteración
    # Este código está indentado (generalmente con 4 espacios)
```

Las partes del bucle son:
*   **`for`**: palabra clave que inicia el bucle.
*   **`item`**: variable que toma el valor de cada elemento de la `secuencia` en cada iteración.
*   **`in`**: palabra clave que dice que objeto se va a iterar.
*   **`secuencia`**: objeto a iterar.
*   **`:`**: final de la declaración `for`.
*   **Bloque de código indentado**: líneas de código indentadas debajo de la línea `for` se ejecutarán en cada iteración del bucle. El bucle acaba cuando acaban las lineas indentadas

````python
edades = [0,10,20,40,60,70]
for edad in edades:
    if edad < 18:
        print(f"Edad {edad}: menor de edad")
    elif edad >= 60:
        print(f"Edad {edad}: adulto mayor")
    else:
        print(f"Edad {edad}: adulto")
````

Para leer un error es importante recordar:
* La flecha verde indica en que línea se cometió el error
* La última línea dice el resumen del error, esta línea es la que se googlea para buscar una solución
